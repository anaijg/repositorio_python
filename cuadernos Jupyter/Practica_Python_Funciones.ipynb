## Ejercicio 6. Devolver un valor con `return`

Define una función llamada `calcular_area_rectangulo(base, altura)` que devuelva el área de un rectángulo mediante `return`.

Después:
1. Guarda en `area1` el resultado de llamar a la función con base `5` y altura `3`.
2. Guarda en `area2` el resultado de llamarla con base `8` y altura `2.5`.
3. Muestra ambas variables.

**Fórmula:** área = base × altura.

````python
# Escribe aquí tu solución
def calcular_area_rectangulo(base, altura):
    return base * altura

area1 = calcular_area_rectangulo(5, 3)
area2 = calcular_area_rectangulo(8, 2.5)
print(area1, area2)
````

## Ejercicio 7. Convertir datos y calcular

## Ejercicio 8. Función que devuelve un resultado booleano

Define una función llamada `es_mayor_de_edad(edad)` que devuelva `True` si la edad es igual o superior a 18 y `False` en caso contrario.

Después, prueba la función con las edades `16`, `18` y `25`, mostrando cada resultado.

**Requisito:** la función debe devolver el resultado con `return`; no basta con mostrarlo mediante `print()` dentro de la función.

````python
# Escribe aquí tu solución
def es_mayor_de_edad(edad):
    return edad > 18
print(es_mayor_de_edad(16))
print(es_mayor_de_edad(18))
print(es_mayor_de_edad(25))
````
