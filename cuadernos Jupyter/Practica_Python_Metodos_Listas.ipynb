## Ejercicio 1. Añadir elementos con `append()`

Crea la lista `tareas = ['estudiar Python', 'revisar apuntes']`.

1. Añade `'hacer ejercicios'` al final usando `append()`.
2. Muestra la lista.
3. Crea la lista `tarea_extra = ['preparar examen', 'entregar práctica']`.
4. Añade `tarea_extra` a `tareas` utilizando `append()`.
5. Muestra la lista y observa qué ocurre con la lista anidada.

## Ejercicio 2. Añadir varios elementos con `extend()`

Crea estas listas:

```python
equipos = ['PC01', 'PC02', 'PC03']
nuevos_equipos = ['PC04', 'PC05', 'PC06']
```

1. Utiliza `extend()` para añadir todos los elementos de `nuevos_equipos` a `equipos`.
2. Muestra `equipos`.
3. Explica mediante un comentario una diferencia entre `append()` y `extend()`.

````python
equipos = ['PC01', 'PC02', 'PC03']
nuevos_equipos = ['PC04', 'PC05', 'PC06']

# Escribe aquí tu solución
equipos.extend(nuevos_equipos)
print(equipos)
# append -> añade la lista al final como lista
# extend -> añade los elementos de la lista al final, como elementos sueltos
````

## Ejercicio 5. Eliminar por índice con `pop()`

Crea la lista `puertos = [21, 22, 25, 53, 80, 443]`.

1. Usa `pop(2)` para eliminar el elemento que ocupa el índice `2` y guarda el valor eliminado en `puerto_eliminado`.

## Ejercicio 7. Ordenar con `sort()` y `sorted()`

Crea la lista `tiempos = [18, 5, 23, 12, 9]`.

1. Utiliza `sort()` para ordenar la lista de menor a mayor.
2. Muestra la lista.
3. Crea una nueva lista llamada `tiempos_descendentes` que contenga los mismos valores en orden descendente usando `sorted(reverse=True)` sobre una copia de la lista o mediante otra lista que crees.
4. Crea la lista `original = [18, 5, 23, 12, 9]`.
5. Utiliza `sorted(original)` y guarda el resultado en `ordenados`.
6. Muestra `original` y `ordenados` para observar que `sorted()` no modifica la lista original.

````python
tiempos = [18, 5, 23, 12, 9]

# Escribe aquí tu solución
tiempos.sort()
print(tiempos)
tiempos_descendentes = sorted(tiempos, reverse=True)
print(tiempos_descendentes)
original = [18, 5, 23, 12, 9]
ordenados = sorted(original)
print(original, ordenados)
````

## Ejercicio 8. Invertir una lista con `reverse()`

## Ejercicio 10. Reto integrador: gestionar paquetes de software

Una lista representa los paquetes instalados en un equipo:

```python
paquetes = ['python', 'git', 'firefox', 'python']
```

Realiza las siguientes operaciones, en este orden:

1. Añade `'vscode'` al final.
2. Añade los paquetes `['docker', 'wireshark']` como elementos individuales con `extend()`.
3. Inserta `'linux'` al principio.
4. Elimina la primera aparición de `'python'` con `remove()`.
5. Busca el índice de `'wireshark'` y guarda el resultado en `indice_wireshark`.
6. Crea una lista ordenada llamada `paquetes_ordenados` utilizando `sorted()` sin modificar `paquetes`.
7. Invierte el orden de `paquetes` con `reverse()`.
8. Muestra `paquetes`, `paquetes_ordenados` e `indice_wireshark`.

**Reto adicional:** crea una lista llamada `paquetes_eliminados` con dos elementos y vacíala usando `clear()`. Muestra el resultado.

````python
paquetes = ['python', 'git', 'firefox', 'python']

# Escribe aquí tu solución
paquetes.append('vscode')
paquetes.extend(['docker', 'wireshark'])
paquetes.insert(0, 'linux')
paquetes.remove('python')
indice_wireshark = paquetes.index('wireshark')
paquetes_ordenados = sorted(paquetes)
paquetes.reverse()
print(paquetes, paquetes_ordenados, indice_wireshark)
````
