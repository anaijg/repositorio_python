## Ejercicio 8. Desempaquetar una lista

Crea la lista `equipo = ['Ana', 'Luis', 'Marta']`.

1. Desempaqueta sus elementos en tres variables: `persona1`, `persona2` y `persona3`.
2. Muestra cada variable en una línea diferente.
3. Crea la lista `datos_programador = ['Sergio', 28, 'Python', 'Madrid']`.
4. Desempaqueta el primer elemento en `nombre` y recoge todos los elementos restantes en una variable llamada `resto`, usando `*`.
5. Muestra `nombre` y `resto`.

**Importante:** el número de variables debe coincidir con el número de elementos, salvo que utilices una variable con `*` para recoger los restantes.

````python
equipo = ['Ana', 'Luis', 'Marta']

# Escribe aquí tu solución
persona1, persona2, persona3 = equipo
print(persona1)
print(persona2)
print(persona3)

datos_programador = ['Sergio', 28, 'Python', 'Madrid']

# Escribe aquí tu solución
nombre, *resto = datos_programador
print(nombre, resto)
````

## Ejercicio 9. Cortar listas (slicing)

Dada la lista:

```python
postres = ['Tarta', 'Galletas', 'Helado', 'Pastel', 'Brownies', 'Fruta']
```

Crea y muestra las siguientes porciones usando slicing:
1. Los elementos desde el índice `1` hasta el `3`, incluyendo el índice `1` pero no el `3`.
2. Los tres primeros elementos.
3. Los dos últimos elementos.
4. Los elementos desde el índice `2` hasta el final.
5. La lista completa mediante una porción.

No modifiques la lista original.

````python
postres = ['Tarta', 'Galletas', 'Helado', 'Pastel', 'Brownies', 'Fruta']

# Escribe aquí tu solución
print(postres[1:3])
print(postres[:3])
print(postres[-2:])
print(postres[:])
````

## Ejercicio 10. Slicing con paso y reto integrador

Parte de esta lista:

```python
numeros = [10, 15, 20, 25, 30, 35, 40, 45, 50, 55]
```

Realiza estas tareas:
1. Obtén una lista con los elementos que ocupan los índices pares (`0`, `2`, `4`, etc.) utilizando slicing con paso.
2. Obtén una lista con los elementos que ocupan los índices impares (`1`, `3`, `5`, etc.) utilizando slicing con paso.
3. Obtén una copia de la lista en orden inverso utilizando slicing.
4. Crea una lista llamada `resumen` con el primer elemento, el último elemento y la longitud de `numeros`.
5. Muestra `resumen`.

**Reto final:** sin cambiar la lista `numeros`, crea una lista llamada `seleccion` que contenga los elementos desde el índice `2` hasta el índice `7` (sin incluirlo), avanzando de dos en dos.

````python
numeros = [10, 15, 20, 25, 30, 35, 40, 45, 50, 55]

# Escribe aquí tu solución
indices_pares = numeros[::2]
print(indices_pares)
indices_impares = numeros[1::2]
print(indices_impares)
inversa = numeros[::-1]
print(inversa)
````
